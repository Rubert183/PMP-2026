```python
import numpy as np

N = 100_000


def primul_zar_par_suma_mai_mare_10():
    cazuri_favorabile = 0
    cazuri_valide = 0
    for _ in range(N):
        x = np.random.randint(1, 7)
        y = np.random.randint(1, 7)
        if x % 2 == 0:
            cazuri_valide += 1
            if x + y >= 10:
                cazuri_favorabile += 1

    return cazuri_favorabile / cazuri_valide


def cel_putin_o_val_para_suma_mai_mare_10():
    cazuri_favorabile = 0
    cazuri_valide = 0
    for _ in range(N):
        x = np.random.randint(1, 7)
        y = np.random.randint(1, 7)

        if x % 2 == 0 or y % 2 == 0:
            cazuri_valide += 1
            if x + y >= 10:
                cazuri_favorabile += 1

    return cazuri_favorabile / cazuri_valide


def suma_8_ambele_pare():
    cazuri_favorabile = 0
    cazuri_valide = 0
    for _ in range(N):
        x = np.random.randint(1, 7)
        y = np.random.randint(1, 7)

        if x + y == 8:
            cazuri_valide += 1
            if x % 2 == 0 and y % 2 == 0:
                cazuri_favorabile += 1

    return cazuri_favorabile / cazuri_valide


print("P(X+Y>=10 | X - par) =", primul_zar_par_suma_mai_mare_10())
print("P(X+Y>=10 | X sau Y - par) =", cel_putin_o_val_para_suma_mai_mare_10())
print("P(X si Y - par | X+Y=8) =", suma_8_ambele_pare())
```
