```python
import numpy as np

N = 1_000_000
bolnav = np.random.random(N) < 0.01

test_pozitiv = np.where(bolnav, np.random.random(N) < 0.95, np.random.random(N) < 0.10)

print("Bolnavi:", bolnav.sum())
print("Sănătosi:", (~bolnav).sum())

print("Teste pozitive:", test_pozitiv.sum())
print("Teste negative:", (~test_pozitiv).sum())

adevarat_pozitiv = bolnav & test_pozitiv
fals_pozitiv = ~bolnav & test_pozitiv

print("Adevarat pozitiv:", adevarat_pozitiv.sum())
print("Fals pozitiv:", fals_pozitiv.sum())

print("P(B | Test pozitiv) =", adevarat_pozitiv.sum() / test_pozitiv.sum())
```
