```python
import numpy as np

rosie = 3
albastru = 4
negru = 2

zar = np.random.randint(1, 7)

if zar in [2, 3, 5]:
    negru += 1
elif zar == 6:
    rosie += 1
else:
    albastru += 1

bila_extrasa = np.random.randint(1, rosie+albastru+negru+1)
print(f"Probabilitatea sa fi scos rosie era {(rosie) / (rosie + negru + albastru)}")

if 1 <= bila_extrasa <= rosie:
    rosie -= 1
    print("Am scos o bila rosie")
elif rosie + 1 <= bila_extrasa <= negru + rosie:
    negru -= 1
    print("Am scos o bile neagra")
else:
    albastru -= 1
    print("Am scos o bila albastra")

print(f"Situatie bile r: {rosie}, n: {negru}, a: {albastru}")
```

```python
Calcularea probalilitatii de a scoate o bila rosie:

1. zarul este un numar prim => P(r) = 3/10 = 0.3
2. zarul este egal cu 6 => P(r) = 4/10 = 0.4
3. zarul are alt numar decat cele mai sus = 3/10 = 0.3

Probalilitatea totala este de P(r) = 0.3 * 3/6 + 0.4 * 1/6 + 0.3 * 2/6 = 0.316
```
